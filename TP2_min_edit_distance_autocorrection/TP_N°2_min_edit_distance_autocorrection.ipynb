```python
def levensh_distance(str1,str2):
    len_str1=len(str1)+1
    len_str2=len(str2)+1
    dp = [[0] * len_str2 for _ in range(len_str1)]
    for i in range(len_str1):
        dp[i][0]=i
    for i in range(len_str2):
        dp[0][i]=i
    for i in range(1, len_str1):
        for j in range(1, len_str2):
            cost=0 if str1[len_str1-1]==str2[len_str2-1] else 2
            dp[i][j]=min(dp[i-1][j]+1,dp[i][j-1]+1,dp[i-1][j-1]+cost)

    return dp[-1][-1]

```

```python
def levensh_distance_with_back(str1, str2):
    len_str1 = len(str1) + 1
    len_str2 = len(str2) + 1
    dp = [[(0, '')] * len_str2 for _ in range(len_str1)]

    for i in range(len_str1):
        dp[i][0] = (i, "delete")
    for i in range(len_str2):
        dp[0][i] = (i, "insert")

    for i in range(1, len_str1):
        for j in range(1, len_str2):
            cost = 0 if str1[i - 1] == str2[j - 1] else 2
            options = [
                (dp[i - 1][j - 1][0] + cost, 'substitute' if cost else 'match'),
                (dp[i - 1][j][0] + 1, 'delete'),
                (dp[i][j - 1][0] + 1, 'insert')
            ]
            dp[i][j] = min(options, key=lambda x: x[0])

    distance = dp[len_str1 - 1][len_str2 - 1][0]

    align1 = []
    align2 = []
    i, j = len_str1 - 1, len_str2 - 1

    while i > 0 or j > 0:
        current_action = dp[i][j][1]
        if current_action == 'match' or current_action == 'substitute':
            align1.append(str1[i - 1])
            align2.append(str2[j - 1])
            i -= 1
            j -= 1
        elif current_action == 'delete':
            align1.append(str1[i - 1])
            align2.append('-')
            i -= 1
        elif current_action == 'insert':
            align1.append('-')
            align2.append(str2[j - 1])
            j -= 1

    align1.reverse()
    align2.reverse()

    return distance, ''.join(align1), ''.join(align2)
```

```python
str1 = "hafid"
str2 = "habinj"
distance, align1, align2 = levensh_distance_with_back(str1, str2)

print(f"Distance d'édition : {distance}")
print(f"Alignement 1 : {align1}")
print(f"Alignement 2 : {align2}")
```

```python
def qwerty_distance(char1, char2, graph):

    if char1 == char2:
        return 0
    if char1 in graph and char2 in graph[char1].values():
        return 0.5
    return 1

print(qwerty_distance('e', 'r', qwerty_graph))
print(qwerty_distance('e', 'z', qwerty_graph))
```

```python
def edit_distance_qwerty(word1, word2, graph):

    len1, len2 = len(word1), len(word2)
    dp = np.zeros((len1 + 1, len2 + 1))

    for i in range(len1 + 1):
        dp[i][0] = i
    for j in range(len2 + 1):
        dp[0][j] = j

    for i in range(1, len1 + 1):
        for j in range(1, len2 + 1):
            cost = qwerty_distance(word1[i - 1], word2[j - 1], graph)
            dp[i][j] = min(dp[i - 1][j] + 1,
                           dp[i][j - 1] + 1,
                           dp[i - 1][j - 1] + cost)

    return dp[len1][len2]

print(edit_distance_qwerty("chat", "chqt", qwerty_graph))
print(edit_distance_qwerty("bonjour", "bonjoul", qwerty_graph))
```

```python
def model_bayes(corpus):

    words=get_vocabulary(corpus)
    total_words=len(words)
    compteurs={}
    for word in words:
        if word in compteurs:
            compteurs[word]=compteurs[word]+1
        else:
            compteurs[word]=1
    proba_of_words={word: count/total_words for word,count in compteurs.items() }
    return proba_of_words
```

```python
def edits1(s):

    letters = string.ascii_lowercase
    splits = [(s[:i], s[i:]) for i in range(len(s) + 1)]

    deletes = [L + R[1:] for L, R in splits if R]
    inserts = [L + c + R for L, R in splits for c in letters]
    substitutes = [L + c + R[1:] for L, R in splits if R for c in letters]

    return set(deletes + inserts + substitutes)

def edits2(s):

    return set(e2 for e1 in edits1(s) for e2 in edits1(e1))

def knownWords(words,vocabulary):

    return set(w for w in words if w in vocabulary)

def candidates(word,vocabulary):

    if word in vocabulary:
        return {word}
    edits1_words = knownWords(edits1(word), vocabulary)
    if edits1_words:
        return edits1_words
    edits2_words = knownWords(edits2(word), vocabulary)
    if edits2_words:
        return edits2_words
    return {word}

def correction(word, vocabulary, word_proba):
    possible_candidates = candidates(word, vocabulary)
    sorted_candidates = sorted(possible_candidates, key=lambda c: word_proba.get(c, 0), reverse=True)
    return sorted_candidates[:3]


```
